#### 12.1 Хистограми

Када немамо пуно података можемо их приказати и низом стубића који се још зове и хистограм. Да бисмо то урадили неопходно је да прво увеземо библиотеку која нам је неопходна.

```python
import matplotlib.pyplot as plt
```

Да бисмо полазали како хистограм изгледа приказаћемо податке о дневном броју састанака једног менаџера током једне недеље.

```python
дани=["пон","уто","сре","чет","пет"]
број_састанака=[5,7,4,6,3]
plt.barh(дани,број_састанака)
plt.title("Број састанака једног менаџера")
plt.show()
plt.close()
```

Сада ћемо још мало средити овај хистограм тако што ћемо му променити димензије и боју у следећим линијама кода:

```python
plt.figure(figsize=(9,5))
plt.bar(дани,број_састанака,color="purple")
plt.title("Број састанака једног менаџера")
plt.show()
plt.close()
```

### 12.2 Приказивање две групе података на истом дијаграму

Понекад је згодно приказати две групе података на једном хистограму. Ево једног примера који то лепо илуструје: За једну недељу су дате минимална и максимална дневна температура у једном граду:

```python
дан=["пон","уто","сре","чет","пет","суб","нед"]
минимална=[15,13,14,16,17,12,13]
максимална=[24,22,23,28,31,27,25]
```

Ове податке ћемо приказати тако што ћемо два пута позвати функцију за цртање хистограма, али водећи рачуна да подаци са већом вредношћу не прекрију оне са мањом вредношћу.

```python
plt.bar(дан,максимална)
plt.bar(дан,минимална)
plt.title("Временска прогноза")
plt.ylabel("Температура (C)")
plt.show()
plt.close()
```

Сада ћемо се још мало играти са дијаграмом. Додаћемо легенду која означава која боја шта значи и променићемо опсег у осе.

```python
plt.bar(дан,максимална,label="максимална температура")
plt.bar(дан,минимална,label="минимална температура")
plt.ylim(10,35)
plt.title("Временска прогноза")
plt.ylabel("Температура (C)")
plt.legend()
plt.show()
plt.close()
```
