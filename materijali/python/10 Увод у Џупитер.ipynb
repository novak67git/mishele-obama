### 10.2 Коришћење променљивих и извршавање Пајтон команди у Џупитеру

У неким ситуацијама нам је много лакше да решимо проблем ако уведемо променљиве. Ово посебно важи када имамо доста података или када имамо бројеве који су велики или када се неки податак користи више пута. 
Пример: Један видео снимак је првог дана на Јутјубу имао 12430793 прегледа. Од тога је 14.23% било из Србије. Израчунати колико је прегледа било из Србије тог дана.
Прво ћемо направити једну променљиву у којој се чува укупан број прегледа.

```python
број_прегледа=12430793
```

Када извршимо ову линију, нема исписа на излазу. Само смо сачували податак у променљивој. Сада рачунамо број прегледа у Србији:

```python
број_прегледа*14.23/100
```

Тако смо добили процену броја прегледа у Србији.
Сада можемо решити мало сложенији задатак: Цена једног производа је на почетку била 5000 динара. Након тога је уследио попуст од 15%, па затим још један попуст од 10%. Након два попуста производ је поново поскупео за 20%. Колика је крајња цена производа?

```python
цена1=5000
цена2=цена1-цена1*15/100
цена3=цена2-цена2*10/100
цена4=цена3+цена3*20/100
цена4
```

Овај програм смо могли написати и на следећи начин:

```python
цена1=5000
цена2=цена1-цена1*15/100
цена3=цена2-цена2*10/100
цена4=цена3+цена3*20/100
print(цена4)
```

У овом случају се програм не извршава већ се само извршава наредба исписа. Ово је згодно користити када имамо неки опис речима при испису или када имамо резултат од више променљивих. Трудићемо се да у већини случајева избегавамо овај начин, односно да га користимо само када је то стварно потребно.

### 10.3 Библиотеке функција

Сви напреднији програмски језици поседују доста унапред уграђених функција које се налазе у библиотекама које се понекад налазе у основном пакету програмског језика, а понекад их је потребно додатно инсталирати. Библиотека која је нама тренутно најзанимљивија јесте math и она ће нам послужити у овом поглављу. У њој се налазе све битније математичке функције.

Пример: Дате су координате неке тачке у координатном систему. Треба наћи растојање те тачке од координатног почетка.

Да бисмо ово решили треба нам функција sqrt, која рачуна квадратни корен неког броја.

```python
from math import sqrt
x=8
y=6
r=sqrt(x*x+y*y)
r
```

Овако изгледа програм када већ унапред знамо параметре. Сада ћемо исти проблем решити за општи случај.

```python
from math import sqrt
x=float(input("Унеси x координату тачке: "))
y=float(input("Унеси y координату тачке: "))
y=6
r=sqrt(x**2+y**2)
print("Растојање тачке од координатног почетка је ",r)
```

Ево још једног примера: Са тастатуре се уноси полупречник лопте. Израчунати њену запремину.
Да бисмо ово решили, треба нам број пи, па ћемо њега узети из библиотеке.

```python
from math import pi
r=float(input("Унесите полупречник лопте: "))
v=r**3*pi
print("Запремина лопте је ",v)
```
