Када извршимо ову команду, не исписује се ништа, али се променљивој која се зове низ додељује низ елемената. Ако желимо да видимо елементе низа, довољно је само да укуцамо низ у новој ћелији.

### 11.2 Приказивање низова података

Помоћу два низа представићемо зависност брзине кретања неког тела од времена. Први низ ће садржати временске тренутке у којима се очитава брзина. Други низ ће садржати брзине у тим временским тренуцима. Битно је само да се позиције у низу одговарајућег времена и брзине поклапају.

```python
време=[0,1,2,3,4,5,6,7]
брзина=[0,1,3,7,6,3,1,0]
```

За цртање графика нам је потребна библиотгека коју ћемо у следећој ћелији учитати.

```python
import matplotlib.pyplot as plt
```

Сада су нам све функције из ове библиотеке доступне, и скраћено ћемо је звати plt. У једној радној свесци довољно је једном учитати библиотеку која нам треба. Сада треба графички представити податке које имамо.

```python
plt.plot(време,брзина)
plt.show()
plt.close()
```

Функција plot нам исцртава график тако да први параметар представља податке са x осе а други са y осе. Функција show нам приказује график, а функција close завршава рад на графику и оставља га ту где јесте. Да би график био комплетан, додајемо наслов, и означавамо шта је на осама на следећи начин:

```python
plt.plot(време,брзина)
plt.title("Зависност брзине од времена")
plt.xlabel("време (с)")
plt.ylabel("брзина(м/с)")
plt.show()
plt.close()
```

#### Задатак 3
За трговца из претходног задатка приказати колико је вредело његово злато током временског периода који је приказан у претходнм задатку.

#### Задатак 4

Аутомобил има почетну брзину од 10 м/с и наредних 10 секунди се креће убрзањем од 2 метра у секунди по секунди. Нацртати график пређеног пута овог аутомобила. 

```python
import matplotlib.pyplot as plt

v0 = 10
a = 2
t = list(range(0, 11))

s = [v0 * x + 0.5 * a * x**2 for x in t]

plt.plot(t, s, marker='o')
plt.xlabel("Vreme (s)")
plt.ylabel("Pređeni put (m)")
plt.title("Grafik pređenog puta automobila")
plt.grid(True)
plt.show()
```

#### Задатак 5
Са апликације коју користите за временску прогнозу очитајте податке о прогнозираној темпереатури за наредна 24 сата и прикажите их графички.

```python
import matplotlib.pyplot as plt

v0 = 10
a = 2
t = list(range(0, 11))

v = [v0 + a * x for x in t]

plt.plot(t, v, marker='o')
plt.xlabel("Vreme (s)")
plt.ylabel("Brzina (m/s)")
plt.title("Grafik brzine automobila")
plt.grid(True)
plt.show()
```
